```python
dbutils.widgets.removeAll()
dbutils.widgets.text("id", "101", "ID")

ID = dbutils.widgets.get("id")
print("Student's id is: ", ID)
   
```

```python
dbutils.widgets.dropdown("course", "Databricks", ["Spark", "PySpark", "Databricks"])
course = dbutils.widgets.get("course")
print("Student's course is: ", course)
```

```python
dbutils.widgets.combobox("roles", "data-science", ["Data-Analyst", "Data-Scientist", "ML engineers"], "ROLES")
role = dbutils.widgets.get("roles")
print("Student's role is: ", role)
```

```python
dbutils.widgets.multiselect("city", "Delhi", ["Delhi", "Mumbai", "Bangalore"], "City")
city = dbutils.widgets.get("city").split(",")
print("Student's choice is: ", city)
```
